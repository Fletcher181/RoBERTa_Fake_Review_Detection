```python
import pandas as pd
import matplotlib.pyplot as plt
import textwrap

# Data
data = {
    "Enhancement Removed": [
        "None (all 4 enhancements)",
        "Cosine LR scheduling",
        "Class-balanced weighted loss",
        "Explicit dropout tuning",
        "Early stopping on F1"
    ],
    "Accuracy": [0.8958, 0.9167, 0.9042, 0.8833, 0.8875],
    "Precision": [0.9060, 0.9168, 0.9042, 0.8860, 0.8702],
    "Recall": [0.9750, 0.9250, 0.9083, 0.9250, 0.8333],
    "F1-Score": [0.9035, 0.9167, 0.9042, 0.8831, 0.8811],
    "FN": [3, 9, 11, 9, 20],
    "Train-Val Gap": ["~0.32", "~0.72", "~0.25", "~0.35", "~0.56"],
    "Stability": ["Good", "Poor", "Good", "Moderate", "Poor"]
}

df = pd.DataFrame(data)

# Light wrap for first column only
df["Enhancement Removed"] = df["Enhancement Removed"].apply(
    lambda x: "\n".join(textwrap.wrap(x, width=20))
)

fig, ax = plt.subplots(figsize=(13, 5.5))  # slightly taller

ax.axis('off')

table = ax.table(
    cellText=df.values,
    colLabels=df.columns,
    cellLoc='center',
    loc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(9)

# KEY: slightly taller rows (not too much)
table.scale(1, 2.2)

# KEY: reduce other columns, keep first column reasonable
col_widths = [
    0.22,  # Enhancement Removed (slightly wider, not too much)
    0.08,  # Accuracy
    0.08,  # Precision
    0.08,  # Recall
    0.08,  # F1-Score
    0.06,  # FN
    0.10,  # Train-Val Gap
    0.10   # Stability
]

for i, width in enumerate(col_widths):
    for (row, col), cell in table.get_celld().items():
        if col == i:
            cell.set_width(width)

# Header styling
for (row, col), cell in table.get_celld().items():
    if row == 0:
        cell.set_text_props(weight='bold')
        cell.set_facecolor('#D9EAF7')

plt.title(
    "Ablation Study: Effect of Removing Enhancements",
    fontsize=13,
    weight='bold',
    pad=12
)

plt.show()
```
